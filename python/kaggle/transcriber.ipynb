```python
import whisper
from whisper.utils import get_writer
from IPython.display import FileLink
import torch
import glob
import os

path = ""
audioFile = "/kaggle/input/datasets/abdulzoha786/bad-uncle/The bad uncle.m4a"
fname,_ = audioFile.split(".")
fileName = f"{fname}.srt"

# 1. Verify GPU is enabled
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

# 2. Load the large-v3 model
model = whisper.load_model("large-v2", device=device)

# 3. Transcribe your audio file (replace "audio.mp3" with your exact filename)

# result = model.transcribe(audioFile, verbose=True, language="en", word_timestamps=True)
result = model.transcribe(audioFile, verbose=True, language="en")

# 4. Write the timestamped subtitles to transaction.srt
srt_writer = get_writer("srt", ".")
# srt_writer(result, fileName, {"max_line_width": 42, "max_line_count": 2})
srt_writer(result, fileName)

print("\nSaved transaction.srt with timestamps!")

# Replace 'your_file.csv' with your actual filename
FileLink(os.listdir()[1])
```
